```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt

# Generate small synthetic dataset (easy to calculate by hand)
X, y = make_regression(
    n_samples=5,
    n_features=2,
    n_informative=2,
    noise=5,
    random_state=42
)

# Use all data for training (small dataset)
X_train, y_train = X, y
X_test, y_test = X, y  # Test on same data for simplicity

# Print the dataset
print("Dataset (X and y):")
print("=" * 40)
for i in range(len(X)):
    print(f"Sample {i+1}: X = [{X[i,0]:7.2f}, {X[i,1]:7.2f}], y = {y[i]:7.2f}")
print()

# Create and train Ridge regression model
ridge = Ridge(alpha=100)
ridge.fit(X_train, y_train)

# Make predictions
y_pred = ridge.predict(X_test)

# Evaluate the model
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Ridge Regression Results")
print("=" * 40)
print(f"Mean Squared Error: {mse:.2f}")
print(f"R² Score: {r2:.4f}")
print(f"\nModel coefficients: {ridge.coef_}")
print(f"Model intercept: {ridge.intercept_:.2f}")

# Plot predictions vs actual values
plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred, alpha=0.6)
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         'r--', lw=2, label='Perfect prediction')
plt.xlabel('Actual Values')
plt.ylabel('Predicted Values')
plt.title('Ridge Regression: Predicted vs Actual')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
