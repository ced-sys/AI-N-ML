```python
import numpy as np
from fractions import Fraction
```

```python
data=[(1,2), (2,8), (3, 10), (4, 11), (5, 10)]
```

```python
x=np.array([point[0] for point in data])
y=np.array([point[1] for point in data])
```

```python
print(f"x values: {x}")
print(f"y values: {y}\n")
```

```python
X=np.column_stack([np.ones(len(x)), x, x**2])
```

```python
print(X)
```

```python
XTX=X.T@X
print(XTX)
```

```python
XTy=X.T @y
print(f"X^T y= {XTy}\n")
```

```python
augmented=np.column_stack([XTX.astype(float), XTy.astype(float)])
print(f"\nAugmented Matrix [XTX | XTy]:\n{augmented}\n")
```

```python
n=len(augmented)
```

```python
for i in range(n):
  #Find pivot (largest value in column)
  max_row=i
  for k in range(i+1, n):
    if abs(augmented[k, i])> abs(augmented[max_row, i]):
      max_row=k

  augmented[[i, max_row]]=augmented[[max_row, i]]

  for k in range(i+1, n):
    if augmented[i, i]!=0:
      factor=augmented[k, i]/augmented[i,i]
      print(f"Row {k+1}: Multiply row {i+1} by {factor:.4f} and subtract")
      augmented[k, :]-=factor*augmented[i,:]
      print(f"Row {k+1} becomes: {augmented[k, :]}\n")

print(f"After forward elimination:\n{augmented}\n")


```

```python
beta=np.zeros(n)

for i in range(n-1, -1, -1):
  beta[i]=augmented[i, n]
  for j in range(i+1, n):
    beta[i]-=augmented[i, j]*beta[j]
  beta[i]/=augmented[i, i]
  print(f"Beta{i}={beta[i]:.6f}")

print()
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
poly_features=PolynomialFeatures(degree=2)
X_poly=poly_features.fit_transform(x.reshape(-1, 1))
```

```python
model = LinearRegression()
model.fit(X_poly, y)
```

```python
beta_sklearn=model.coef_
intercept_sklearn=model.intercept_
```

```python
print(f"\nBeta (intercept)={intercept_sklearn:.6f}")
print(f"\nBeta (linear)={beta_sklearn[1]:.6f}")
print(f"\nBeta (quadratic)={beta_sklearn[2]:.6f}")
```
