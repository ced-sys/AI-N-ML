```python
# ================================
# Track 6: Climate Change vs Forest Cover Change
# Complete Google Colab Pipeline for Observable Integration
# ================================

# ================================
# 1. Setup & Install Dependencies
# ================================
!pip install earthengine-api geemap folium geopandas scikit-learn plotly seaborn pandas numpy

import ee
import geemap
import pandas as pd
import numpy as np
import geopandas as gpd
from datetime import datetime
import json
import os

# Authenticate Earth Engine
print("🔐 Authenticating Earth Engine...")
ee.Authenticate()
ee.Initialize(project='ee-victorcedric665')
print("✅ Earth Engine authenticated and initialized!")

# ================================
# 2. Configuration & Setup
# ================================

class Track6DataPipeline:
    """Complete data pipeline for Track 6 analysis"""

    def __init__(self, country='Kenya', admin_level=1):
        """
        Initialize pipeline with country and admin level

        Args:
            country: Country name (e.g., 'Kenya', 'Uganda', 'Tanzania')
            admin_level: 0=country, 1=state/province, 2=district
        """
        self.country = country
        self.admin_level = admin_level
        self.exports = {}

        print(f"🌍 Initialized Track 6 pipeline for {country} (Admin Level {admin_level})")

    def load_admin_boundaries(self):
        """Load administrative boundaries"""
        print(f"📍 Loading administrative boundaries for {self.country}...")

        # Load GAUL boundaries
        all_admin = ee.FeatureCollection(f"FAO/GAUL/2015/level{self.admin_level}")

        # Filter by country
        self.admin = all_admin.filter(ee.Filter.eq('ADM0_NAME', self.country))

        count = self.admin.size().getInfo()
        print(f"✅ Loaded {count} administrative units")

        return self.admin

    def compute_forest_metrics(self):
        """Compute forest cover and loss metrics (2000-2023)"""
        print("🌲 Computing forest metrics...")

        # Load Hansen Global Forest Change
        hansen = ee.Image("UMD/hansen/global_forest_change_2024_v1_12")

        # Forest cover 2000 (>30% threshold)
        tree_cover_2000 = hansen.select('treecover2000')
        self.forest_mask_2000 = tree_cover_2000.gte(30)

        # Loss year (1-23 = 2001-2023)
        self.loss_year = hansen.select('lossyear')

        # Forest gain
        self.forest_gain = hansen.select('gain')

        # Pixel area in hectares
        pixel_area = ee.Image.pixelArea().divide(10000)

        # Forest area 2000
        forest_area_2000 = self.forest_mask_2000.multiply(pixel_area)

        def compute_admin_stats(feature):
            """Extract forest stats for each admin unit"""
            admin_name = feature.get(f'ADM{self.admin_level}_NAME')

            # Forest area in 2000
            forest_2000_ha = forest_area_2000.reduceRegion(
                reducer=ee.Reducer.sum(),
                geometry=feature.geometry(),
                scale=30,
                maxPixels=1e12
            ).get('treecover2000')

            # Total loss area (all years)
            loss_mask = self.loss_year.gte(1)
            total_loss_ha = loss_mask.multiply(pixel_area).reduceRegion(
                reducer=ee.Reducer.sum(),
                geometry=feature.geometry(),
                scale=30,
                maxPixels=1e12
            ).get('lossyear')

            # Loss by year (2001-2023)
            loss_by_year = {}
            for year in range(1, 24):  # 1-23
                year_mask = self.loss_year.eq(year)
                year_loss = year_mask.multiply(pixel_area).reduceRegion(
                    reducer=ee.Reducer.sum(),
                    geometry=feature.geometry(),
                    scale=30,
                    maxPixels=1e12
                ).get('lossyear')
                loss_by_year[f'loss_{2000+year}'] = year_loss

            # Forest gain area
            gain_ha = self.forest_gain.multiply(pixel_area).reduceRegion(
                reducer=ee.Reducer.sum(),
                geometry=feature.geometry(),
                scale=30,
                maxPixels=1e12
            ).get('gain')

            # Calculate forest loss percentage
            loss_percent = ee.Algorithms.If(
                ee.Number(forest_2000_ha).gt(0),
                ee.Number(total_loss_ha).divide(forest_2000_ha).multiply(100),
                0
            )

            # Create properties dictionary
            props = {
                'admin_name': admin_name,
                'forest_area_2000_ha': forest_2000_ha,
                'total_loss_ha': total_loss_ha,
                'loss_percent': loss_percent,
                'gain_ha': gain_ha
            }
            props.update(loss_by_year)

            return feature.set(props)

        # Apply to all admin units
        self.forest_stats = self.admin.map(compute_admin_stats)

        print("✅ Forest metrics computed")
        return self.forest_stats

    def compute_climate_hazards(self):
        """Compute climate hazard metrics"""
        print("🌡️ Computing climate hazard metrics...")

        def compute_hazard_stats(feature):
            """Extract climate hazard stats for each admin unit"""

            # Temperature - ERA5 Monthly (2000-2023)
            era5 = ee.ImageCollection("ECMWF/ERA5/MONTHLY") \
                     .select('mean_2m_air_temperature') \
                     .filterDate('2000-01-01', '2023-12-31')

            # Convert to Celsius and compute mean
            temp_celsius = era5.map(lambda img: img.subtract(273.15))
            mean_temp = temp_celsius.mean()
            max_temp = temp_celsius.max()

            # Temperature trend (linear regression)
            def add_time_band(img):
                date = ee.Date(img.get('system:time_start'))
                years = date.difference(ee.Date('2000-01-01'), 'year')
                return img.addBands(ee.Image(years).rename('t').float())

            temp_with_time = temp_celsius.map(add_time_band)
            temp_trend = temp_with_time.select(['t', 'mean_2m_air_temperature']) \
                                       .reduce(ee.Reducer.linearFit())

            # Drought index - TerraClimate PDSI (2000-2023)
            terraclimate = ee.ImageCollection("IDAHO_EPSCOR/TERRACLIMATE") \
                            .select('pdsi') \
                            .filterDate('2000-01-01', '2023-12-31')

            mean_pdsi = terraclimate.mean()
            min_pdsi = terraclimate.min()  # Most severe drought

            # Precipitation - CHIRPS (2000-2023)
            chirps = ee.ImageCollection("UCSB-CHG/CHIRPS/DAILY") \
                       .filterDate('2000-01-01', '2023-12-31')

            # Annual precipitation
            total_precip = chirps.sum()

            # Fire frequency - MODIS Burned Area (2001-2023)
            modis_fire = ee.ImageCollection("MODIS/006/MCD64A1") \
                          .select('BurnDate') \
                          .filterDate('2001-01-01', '2023-12-31')

            fire_count = modis_fire.count()

            # Reduce all metrics to admin geometry
            geometry = feature.geometry()

            temp_stats = mean_temp.reduceRegion(
                reducer=ee.Reducer.mean(),
                geometry=geometry,
                scale=11132,
                maxPixels=1e12
            )

            temp_trend_stats = temp_trend.select('scale').reduceRegion(
                reducer=ee.Reducer.mean(),
                geometry=geometry,
                scale=11132,
                maxPixels=1e12
            )

            pdsi_stats = mean_pdsi.reduceRegion(
                reducer=ee.Reducer.mean(),
                geometry=geometry,
                scale=4638,
                maxPixels=1e12
            )

            precip_stats = total_precip.reduceRegion(
                reducer=ee.Reducer.mean(),
                geometry=geometry,
                scale=5566,
                maxPixels=1e12
            )

            fire_stats = fire_count.reduceRegion(
                reducer=ee.Reducer.mean(),
                geometry=geometry,
                scale=463,
                maxPixels=1e12
            )

            return feature.set({
                'mean_temperature_c': temp_stats.get('mean_2m_air_temperature'),
                'temp_trend_c_per_year': temp_trend_stats.get('scale'),
                'mean_pdsi': pdsi_stats.get('pdsi'),
                'total_precipitation_mm': precip_stats.get('precipitation'),
                'fire_frequency': fire_stats.get('BurnDate')
            })

        # Apply to forest stats
        self.hazard_stats = self.forest_stats.map(compute_hazard_stats)

        print("✅ Climate hazard metrics computed")
        return self.hazard_stats

    def compute_protected_areas_overlap(self):
        """Compute forest loss within protected areas"""
        print("🛡️ Computing protected areas vulnerability...")

        # Load WDPA
        wdpa = ee.FeatureCollection("WCMC/WDPA/current/polygons")

        # Filter to country
        country_wdpa = wdpa.filterBounds(self.admin.geometry())

        # Pixel area
        pixel_area = ee.Image.pixelArea().divide(10000)

        def compute_pa_loss(pa_feature):
            """Compute forest loss within each protected area"""
            pa_name = pa_feature.get('NAME')
            pa_iucn = pa_feature.get('IUCN_CAT')

            # Forest area in 2000 within PA
            forest_2000_pa = self.forest_mask_2000.multiply(pixel_area).reduceRegion(
                reducer=ee.Reducer.sum(),
                geometry=pa_feature.geometry(),
                scale=30,
                maxPixels=1e12
            ).get('treecover2000')

            # Total loss within PA
            loss_mask = self.loss_year.gte(1)
            loss_pa = loss_mask.multiply(pixel_area).reduceRegion(
                reducer=ee.Reducer.sum(),
                geometry=pa_feature.geometry(),
                scale=30,
                maxPixels=1e12
            ).get('lossyear')

            # Loss percentage
            loss_percent = ee.Algorithms.If(
                ee.Number(forest_2000_pa).gt(0),
                ee.Number(loss_pa).divide(forest_2000_pa).multiply(100),
                0
            )

            return pa_feature.set({
                'pa_name': pa_name,
                'pa_iucn_category': pa_iucn,
                'forest_2000_pa_ha': forest_2000_pa,
                'loss_pa_ha': loss_pa,
                'loss_percent_pa': loss_percent
            })

        self.pa_stats = country_wdpa.map(compute_pa_loss)

        print("✅ Protected areas vulnerability computed")
        return self.pa_stats

    def compute_reforestation_suitability(self):
        """Compute reforestation suitability scores"""
        print("🌱 Computing reforestation suitability...")

        # Load additional datasets
        elevation = ee.Image("USGS/SRTMGL1_003")
        slope = ee.Terrain.slope(elevation)

        # Population density (WorldPop)
        population = ee.ImageCollection("WorldPop/GP/100m/pop") \
                      .filterDate('2020-01-01', '2020-12-31') \
                      .mean()

        # Land cover (MODIS)
        landcover = ee.ImageCollection("MODIS/006/MCD12Q1") \
                     .filterDate('2020-01-01', '2020-12-31') \
                     .first() \
                     .select('LC_Type1')

        def compute_suitability_stats(feature):
            """Compute suitability metrics for each admin unit"""
            geometry = feature.geometry()

            # Elevation stats
            elev_stats = elevation.reduceRegion(
                reducer=ee.Reducer.mean(),
                geometry=geometry,
                scale=90,
                maxPixels=1e12
            )

            # Slope stats
            slope_stats = slope.reduceRegion(
                reducer=ee.Reducer.mean(),
                geometry=geometry,
                scale=90,
                maxPixels=1e12
            )

            # Population density
            pop_stats = population.reduceRegion(
                reducer=ee.Reducer.mean(),
                geometry=geometry,
                scale=100,
                maxPixels=1e12
            )

            # Suitable land cover classes (grassland, shrubland, etc.)
            # Classes: 10=Grassland, 12=Cropland, 16=Barren
            suitable_lc = landcover.eq(10).Or(landcover.eq(16))
            suitable_area = suitable_lc.multiply(ee.Image.pixelArea().divide(10000))

            lc_stats = suitable_area.reduceRegion(
                reducer=ee.Reducer.sum(),
                geometry=geometry,
                scale=500,
                maxPixels=1e12
            )

            # Compute composite suitability score
            # Good conditions: moderate elevation (500-2000m), low slope (<15°), low population
            elevation_score = ee.Number(elev_stats.get('elevation')).subtract(500).abs().divide(1500)
            slope_score = ee.Number(slope_stats.get('slope')).divide(15)
            pop_score = ee.Number(pop_stats.get('population')).divide(100)

            # Normalize scores (0-1, lower is better)
            elevation_score = ee.Number(1).subtract(elevation_score.min(1))
            slope_score = ee.Number(1).subtract(slope_score.min(1))
            pop_score = ee.Number(1).subtract(pop_score.min(1))

            # Weighted composite (higher is more suitable)
            suitability_score = elevation_score.multiply(0.3) \
                              .add(slope_score.multiply(0.4)) \
                              .add(pop_score.multiply(0.3)) \
                              .multiply(100)

            return feature.set({
                'mean_elevation_m': elev_stats.get('elevation'),
                'mean_slope_deg': slope_stats.get('slope'),
                'population_density': pop_stats.get('population'),
                'suitable_land_ha': lc_stats.get('LC_Type1'),
                'reforestation_suitability_score': suitability_score
            })

        self.suitability_stats = self.hazard_stats.map(compute_suitability_stats)

        print("✅ Reforestation suitability computed")
        return self.suitability_stats

    def compute_cobenefits(self):
        """Compute carbon and biodiversity co-benefits"""
        print("🌿 Computing co-benefits (carbon & biodiversity)...")

        # Above-ground biomass (NASA)
        biomass = ee.ImageCollection("NASA/ORNL/biomass_carbon_density/v1") \
                   .select('agb') \
                   .first()

        def compute_cobenefit_stats(feature):
            """Compute co-benefit metrics for each admin unit"""
            geometry = feature.geometry()

            # Carbon density (Mg/ha)
            carbon_stats = biomass.reduceRegion(
                reducer=ee.Reducer.mean(),
                geometry=geometry,
                scale=300,
                maxPixels=1e12
            )

            # Calculate potential carbon sequestration
            # Assume 50 Mg C/ha for restored forest (conservative estimate)
            suitable_land = feature.get('suitable_land_ha')
            carbon_potential = ee.Number(suitable_land).multiply(50)

            # Biodiversity score (proximity to protected areas)
            wdpa = ee.FeatureCollection("WCMC/WDPA/current/polygons")
            pa_distance = wdpa.filterBounds(geometry).distance(maxError=1000)

            # Get the distance value from the reduceRegion result
            distance_value = pa_distance.reduceRegion(
                reducer=ee.Reducer.mean(),
                geometry=geometry,
                scale=1000,
                maxPixels=1e12
            ).get('distance')

            # Calculate biodiversity score, handling potential null distance values
            biodiversity_score = ee.Algorithms.If(
                distance_value,  # Check if distance_value is not null
                ee.Number(100).subtract(ee.Number(distance_value).divide(1000)).max(0),
                0 # Default score if distance is null
            )


            return feature.set({
                'mean_carbon_density_mg_ha': carbon_stats.get('agb'),
                'carbon_sequestration_potential_mg': carbon_potential,
                'biodiversity_proximity_score': biodiversity_score
            })

        self.cobenefit_stats = self.suitability_stats.map(compute_cobenefit_stats)

        print("✅ Co-benefits computed")
        return self.cobenefit_stats

    def export_to_csv(self, description='track6_analysis'):
        """Export results to CSV"""
        print(f"📤 Exporting results to CSV...")

        # Convert to pandas DataFrame using an alternative method
        try:
            # Convert FeatureCollection to list of dictionaries and then to DataFrame
            feature_list = self.cobenefit_stats.getInfo()['features']
            data = pd.DataFrame([f['properties'] for f in feature_list])

            # Clean data
            data = data.dropna(subset=['admin_name'])

            # Save to CSV
            filename = f"{description}_{self.country.lower()}_admin{self.admin_level}.csv"
            data.to_csv(filename, index=False)

            print(f"✅ Exported to {filename}")
            print(f"   Shape: {data.shape}")
            print(f"   Columns: {list(data.columns)}")

            self.exports['csv'] = filename
            return data

        except Exception as e:
            print(f"❌ Export failed: {e}")
            print("   Trying alternative export method via Drive...")

            # Alternative: Export via Earth Engine task
            task = ee.batch.Export.table.toDrive(
                collection=self.cobenefit_stats,
                description=description,
                fileFormat='CSV'
            )
            task.start()
            print(f"   Task started: {description}")
            print(f"   Check your Google Drive for the export")

            return None

    def export_to_geojson(self, description='track6_analysis'):
        """Export results to GeoJSON"""
        print(f"📤 Exporting results to GeoJSON...")

        try:
            # Convert FeatureCollection to GeoJSON
            geojson = self.cobenefit_stats.getInfo()

            # Save to file
            filename = f"{description}_{self.country.lower()}_admin{self.admin_level}.geojson"
            with open(filename, 'w') as f:
                json.dump(geojson, f)

            print(f"✅ Exported to {filename}")

            self.exports['geojson'] = filename
            return geojson

        except Exception as e:
            print(f"❌ Export failed: {e}")
            print("   Trying alternative export method via Drive...")

            # Alternative: Export via Earth Engine task
            task = ee.batch.Export.table.toDrive(
                collection=self.cobenefit_stats,
                description=f"{description}_geojson",
                fileFormat='GeoJSON'
            )
            task.start()
            print(f"   Task started: {description}_geojson")

            return None


    def export_protected_areas(self, description='track6_protected_areas'):
        """Export protected areas analysis"""
        print(f"📤 Exporting protected areas analysis...")

        try:
             # Convert FeatureCollection to list of dictionaries and then to DataFrame
            pa_feature_list = self.pa_stats.getInfo()['features']
            pa_data = pd.DataFrame([f['properties'] for f in pa_feature_list])

            pa_data = pa_data.dropna(subset=['pa_name'])

            filename = f"{description}_{self.country.lower()}.csv"
            pa_data.to_csv(filename, index=False)

            print(f"✅ Exported to {filename}")
            print(f"   Protected areas analyzed: {len(pa_data)}")

            self.exports['protected_areas'] = filename
            return pa_data

        except Exception as e:
            print(f"❌ Export failed: {e}")
            print("   Trying alternative export method via Drive...")
             # Alternative: Export via Earth Engine task
            task = ee.batch.Export.table.toDrive(
                collection=self.pa_stats,
                description=description,
                fileFormat='CSV'
            )
            task.start()
            print(f"   Task started: {description}")
            print(f"   Check your Google Drive for the export")
            return None

    def generate_summary_statistics(self, data):
        """Generate summary statistics for storytelling"""
        print("\n" + "="*60)
        print(f"📊 SUMMARY STATISTICS - {self.country.upper()}")
        print("="*60)

        if data is not None and len(data) > 0:
            # Forest metrics
            total_forest_2000 = data['forest_area_2000_ha'].sum()
            total_loss = data['total_loss_ha'].sum()
            avg_loss_pct = data['loss_percent'].mean()

            print(f"\n🌲 FOREST METRICS:")
            print(f"   Total forest area (2000): {total_forest_2000:,.0f} ha")
            print(f"   Total forest loss (2001-2023): {total_loss:,.0f} ha")
            print(f"   Average loss percentage: {avg_loss_pct:.2f}%")
            print(f"   Loss rate: {(total_loss/total_forest_2000*100):.2f}% of 2000 forest")

            # Climate hazards
            avg_temp = data['mean_temperature_c'].mean()
            avg_temp_trend = data['temp_trend_c_per_year'].mean()
            avg_pdsi = data['mean_pdsi'].mean()

            print(f"\n🌡️ CLIMATE HAZARDS:")
            print(f"   Average temperature (2000-2023): {avg_temp:.1f}°C")
            print(f"   Temperature trend: {avg_temp_trend:.3f}°C/year")
            print(f"   Average drought index (PDSI): {avg_pdsi:.2f}")
            print(f"   Average fire frequency: {data['fire_frequency'].mean():.1f} events")

            # Reforestation potential
            total_suitable = data['suitable_land_ha'].sum()
            avg_suitability = data['reforestation_suitability_score'].mean()
            total_carbon_potential = data['carbon_sequestration_potential_mg'].sum()

            print(f"\n🌱 REFORESTATION POTENTIAL:")
            print(f"   Total suitable land: {total_suitable:,.0f} ha")
            print(f"   Average suitability score: {avg_suitability:.1f}/100")
            print(f"   Carbon sequestration potential: {total_carbon_potential:,.0f} Mg")

            # Top priority areas
            print(f"\n🎯 TOP 5 PRIORITY AREAS FOR RESTORATION:")
            top_areas = data.nlargest(5, 'reforestation_suitability_score')[
                ['admin_name', 'loss_percent', 'reforestation_suitability_score',
                 'suitable_land_ha', 'carbon_sequestration_potential_mg']
            ]
            for idx, row in top_areas.iterrows():
                print(f"   {row['admin_name']}:")
                print(f"      • Loss: {row['loss_percent']:.1f}%")
                print(f"      • Suitability: {row['reforestation_suitability_score']:.1f}/100")
                print(f"      • Available land: {row['suitable_land_ha']:,.0f} ha")
                print(f"      • Carbon potential: {row['carbon_sequestration_potential_mg']:,.0f} Mg")

            print("\n" + "="*60)
        else:
            print("   ⚠️ No data available for summary")

# ================================
# 3. Run Complete Pipeline
# ================================

print("\n" + "="*60)
print("🚀 STARTING TRACK 6 DATA PIPELINE")
print("="*60 + "\n")

# Initialize pipeline (change country as needed)
pipeline = Track6DataPipeline(country='Kenya', admin_level=1)

# Step 1: Load administrative boundaries
pipeline.load_admin_boundaries()

# Step 2: Compute forest metrics
pipeline.compute_forest_metrics()

# Step 3: Compute climate hazards
pipeline.compute_climate_hazards()

# Step 4: Compute protected areas overlap
pipeline.compute_protected_areas_overlap()

# Step 5: Compute reforestation suitability
pipeline.compute_reforestation_suitability()

# Step 6: Compute co-benefits
pipeline.compute_cobenefits()

# Step 7: Export results
print("\n" + "="*60)
print("📤 EXPORTING RESULTS")
print("="*60 + "\n")

# Export main analysis to CSV
main_data = pipeline.export_to_csv(description='track6_main_analysis')

# Export to GeoJSON for mapping
pipeline.export_to_geojson(description='track6_main_analysis')

# Export protected areas analysis
pa_data = pipeline.export_protected_areas(description='track6_protected_areas')

# Step 8: Generate summary statistics
if main_data is not None:
    pipeline.generate_summary_statistics(main_data)

# ================================
# 4. Create Quick Visualization
# ================================

print("\n" + "="*60)
print("🗺️ CREATING PREVIEW MAP")
print("="*60 + "\n")

if main_data is not None and len(main_data) > 0:
    import matplotlib.pyplot as plt
    import seaborn as sns

    # Set style
    sns.set_style("whitegrid")

    # Create figure with subplots
    fig, axes = plt.subplots(2, 2, figsize=(15, 12))
    fig.suptitle(f'Track 6 Analysis Preview - {pipeline.country}', fontsize=16, fontweight='bold')

    # 1. Forest loss percentage by admin unit
    ax1 = axes[0, 0]
    top_loss = main_data.nlargest(10, 'loss_percent')
    ax1.barh(top_loss['admin_name'], top_loss['loss_percent'], color='#d62728')
    ax1.set_xlabel('Forest Loss (%)')
    ax1.set_title('Top 10 Areas by Forest Loss Percentage')
    ax1.invert_yaxis()

    # 2. Reforestation suitability scores
    ax2 = axes[0, 1]
    top_suitable = main_data.nlargest(10, 'reforestation_suitability_score')
    ax2.barh(top_suitable['admin_name'], top_suitable['reforestation_suitability_score'], color='#2ca02c')
    ax2.set_xlabel('Suitability Score')
    ax2.set_title('Top 10 Areas for Reforestation')
    ax2.invert_yaxis()

    # 3. Forest loss vs temperature trend
    ax3 = axes[1, 0]
    ax3.scatter(main_data['temp_trend_c_per_year'], main_data['loss_percent'],
                alpha=0.6, s=100, c='#ff7f0e')
    ax3.set_xlabel('Temperature Trend (°C/year)')
    ax3.set_ylabel('Forest Loss (%)')
    ax3.set_title('Forest Loss vs Climate Change')
    ax3.grid(True, alpha=0.3)

    # 4. Carbon sequestration potential
    ax4 = axes[1, 1]
    top_carbon = main_data.nlargest(10, 'carbon_sequestration_potential_mg')
    ax4.barh(top_carbon['admin_name'], top_carbon['carbon_sequestration_potential_mg']/1000, color='#9467bd')
    ax4.set_xlabel('Carbon Potential (1000 Mg)')
    ax4.set_title('Top 10 Areas by Carbon Sequestration Potential')
    ax4.invert_yaxis()

    plt.tight_layout()
    plt.savefig(f'track6_preview_{pipeline.country.lower()}.png', dpi=300, bbox_inches='tight')
    print("✅ Preview chart saved")
    plt.show()

print("\n" + "="*60)
print("✅ PIPELINE COMPLETE!")
print("="*60)
print(f"\n📁 Exported files:")
for key, filename in pipeline.exports.items():
    print(f"   • {key}: {filename}")

print("\n📝 Next steps:")
print("   1. Download the CSV and GeoJSON files from this Colab")
print("   2. Upload them to your Observable notebook using FileAttachment")
print("   3. Use the Observable code snippets to create interactive visualizations")
print("   4. Build the narrative story following the Track 6 rubric")
print("   5. Export to HTML and submit to Zindi")

print("\n🎯 Ready for Observable integration!")
```
